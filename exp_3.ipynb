```python
import json
from datetime import datetime

def generate_report(case_id, examiner, findings, output_file="report.md"):
    report = f"# Digital Forensics Investigation Report\n\n"
    report += f"**Case ID:** `{case_id}`\n"
    report += f"**Lead Examiner:** {examiner}\n"
    report += f"**Report Generated:** {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n"
    report += "-" * 50 + "\n\n"

    report += "## 1. Executive Summary\n"
    report += f"{findings.get('summary', 'No summary provided.')}\n\n"

    report += "## 2. Evidence Analyzed\n"
    for evidence in findings.get('evidence', []):
        report += f"- **File:** {evidence.get('filename')}\n"
        report += f"  - SHA-256: `{evidence.get('hash')}`\n"

    report += "\n## 3. Key Findings\n"
    for item in findings.get('key_findings', []):
        report += f"- {item}\n"

    report += "\n## 4. Conclusion\n"
    report += f"{findings.get('conclusion', 'Pending investigation.')}\n"

    with open(output_file, 'w') as f:
        f.write(report)

    print(f"[+] Forensic report successfully generated: {output_file}")
```
