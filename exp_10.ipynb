```python
import email

def detect_email_spoofing(raw_email_data):
    msg = email.message_from_string(raw_email_data)

    from_header = msg.get('From', '')
    return_path = msg.get('Return-Path', '')
    reply_to = msg.get('Reply-To', '')

    print("=== Email Header Analysis ===")
    print(f"From:        {from_header}")
    print(f"Return-Path: {return_path}")
    print(f"Reply-To:    {reply_to}")

    clean_return = return_path.strip('<>')

    if clean_return and clean_return not in from_header:
        print("[!] SUSPICIOUS: Return-Path does not match 'From' address.")
    if reply_to and reply_to not in from_header:
        print("[!] SUSPICIOUS: Reply-To is directed to a different address.")

    auth_results = msg.get('Authentication-Results', '')
    if 'spf=fail' in auth_results.lower() or 'dkim=fail' in auth_results.lower():
        print("[!] ALERT: SPF or DKIM validation failed.")
```
