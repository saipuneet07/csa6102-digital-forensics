```python
from collections import defaultdict

def detect_port_scan(network_logs, threshold=10):
    ip_ports_touched = defaultdict(set)

    for log in network_logs:
        src_ip = log.get("src_ip")
        dst_port = log.get("dst_port")
        if src_ip and dst_port:
            ip_ports_touched[src_ip].add(dst_port)

    print("=== Port Scan Detection ===")
    for ip, ports in ip_ports_touched.items():
        if len(ports) >= threshold:
            print(f"[!] SCAN DETECTED: {ip} accessed {len(ports)} unique ports.")
```
