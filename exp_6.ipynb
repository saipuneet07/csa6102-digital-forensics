```python
# Common magic numbers (file signatures)
MAGIC_NUMBERS = {
    b'\xFF\xD8\xFF': 'JPEG Image',
    b'%PDF': 'PDF Document',
    b'PK\x03\x04': 'ZIP Archive',
    b'\x89PNG\x0D\x0A\x1A\x0A': 'PNG Image',
    b'MZ': 'Windows Executable (EXE)',
    b'\x7FELF': 'Linux Executable (ELF)'
}

def identify_file_type(filepath):
    """Identifies true file type regardless of extension."""
    try:
        with open(filepath, 'rb') as f:
            header = f.read(8) # Read first 8 bytes

        for magic, filetype in MAGIC_NUMBERS.items():
            if header.startswith(magic):
                return filetype
        return "Unknown or Unsupported Type"
    except FileNotFoundError:
        return "File not found."

# Example usage:
# print(identify_file_type("suspicious_file.xyz"))
```
